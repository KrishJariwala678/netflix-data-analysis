```python
df['director'] = df['director'].fillna('Director not specified')
df.head()
```

```python
df['director'].isnull().sum()
```

Therefore we can see that there are no missing values in director column

```python
director_df = df[['director']].copy()
director_df['director'] = director_df['director'].str.split(',')
director_df = director_df.explode('director')
director_df['director'] = director_df['director'].str.strip()
```

```python
print(director_df) # you can see the type ,its a dataframe
```

```python
director = director_df.groupby(['director']).size().reset_index(name = 'Total Counts')
print(director)
```

```python
# Cleaning the data
director = director[director.director != 'Director not specified']
print(director)
```

```python
# to find the top 5 director
director = director.sort_values(by = ['Total Counts'] , ascending = False)
print(director)
```

```python
top5Director = director.head()
print(top5Director)
```

```python
top5Director = top5Director.sort_values(by = ['Total Counts'] , ascending = True)
```

```python
barChart = px.bar(top5Director , x = 'Total Counts' ,y = 'director' , title = 'Top 5 Directors on Netflix')
barChart.show()
```

```python
df['cast'] = df['cast'].fillna('Cast not specified')
df.cast
```

```python
cast_df = pd.DataFrame()
cast_df = df['cast'].str.split(',',expand = True).stack()
cast_df = cast_df.to_frame()
cast_df.columns = ['Actor']
actors = cast_df.groupby(['Actor']).size().reset_index(name = 'Total count')
actors = actors[actors.Actor != 'Cast not specified']
actors.sort_values(by = ['Total count'] , ascending = False, inplace = True)
top5Actors = actors.head().copy()
top5Actors.sort_values(by = ['Total count'], ascending = True, inplace = True)
top5Actors = top5Actors.reset_index(drop=True)
top5Actors
```

```python
barChart2 = px.bar(top5Actors , x = 'Total count' , y = 'Actor' ,title = 'Top 5 actors on netflix')
barChart2.show()
```

Sentiment analysis of netflix content

```python
df.columns
df.head()
```

```python
df3 = df[['release_year','description']]
df3 = df3.rename(columns = {"release_year" : "Release Year" , "description" : "Description"})
for index, row in df3.iterrows():
  d=row['Description']
  testimonial = TextBlob(d)
  p = testimonial.sentiment.polarity
  if p==0:
    sent = 'Neutral'
  elif p>0:
    sent = 'Positive'
  else:
    sent = 'Negative'
  df3.loc[[index, 2], 'Sentiment']=sent

df3 = df3.groupby(['Release Year', 'Sentiment']).size().reset_index(name = 'Total Count')

df3 = df3[df3['Release Year']>2005]
barGraph = px.bar(df3, x="Release Year", y="Total Count", color = "Sentiment", title = "Sentiment Analysis of Content on Netflix")
barGraph.show()

```
